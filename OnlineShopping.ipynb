```python
#Gender vs Avg_Price
#To check if spending behavior differs by gender.
plt.figure(figsize=(8,5))
sns.boxplot(x='Gender', y='Avg_Price', data=df)
plt.title('Avg Price by Gender')
plt.show()
```

Insight:
Shows whether male or female customers tend to spend more per transaction and highlights price variability across genders.

```python
#Product_Category vs Quantity
#To understand which categories are bought in higher quantities.
plt.figure(figsize=(12,6))
sns.boxplot(x='Product_Category', y='Quantity', data=df)
plt.xticks(rotation=90)
plt.title('Quantity Purchased per Product Category')
plt.show()
```

Insight:
Identifies categories with bulk purchases and categories mostly bought as single items.

```python
#Coupon_Status vs Avg_Price
#To check whether coupons influence spending amount.
plt.figure(figsize=(8,5))
sns.boxplot(x='Coupon_Status', y='Avg_Price', data=df)
plt.title('Avg Price vs Coupon Status')
plt.show()
```

Insight:
Helps determine if customers spend more when coupons are used.

```python
#Tenure_Months vs Quantity
#To analyze if loyal customers buy more items.
plt.figure(figsize=(8,5))
sns.scatterplot(x='Tenure_Months', y='Quantity', data=df)
plt.title('Tenure vs Quantity Purchased')
plt.show()
```

Insight:
Shows whether long-term customers tend to purchase more per transaction.

MULTIVARIATE ANALYSIS

```python
#Avg_Price vs Quantity by Coupon_Status
#To see combined effect of price, quantity, and coupon usage.
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Avg_Price',
    y='Quantity',
    hue='Coupon_Status',
    data=df
)
plt.title('Avg Price vs Quantity by Coupon Status')
plt.show()
```

Insight:
Shows whether coupon users buy higher quantities or higher-priced products compared to non-users.

```python
#Avg_Price vs Tenure_Months by Gender
#To analyze spending behavior based on loyalty and gender.
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Tenure_Months',
    y='Avg_Price',
    hue='Gender',
    data=df
)
plt.title('Tenure vs Avg Price by Gender')
plt.show()
```

Insight:
Reveals whether long-term male or female customers spend more.

```python
#Correlation Heatmap (Numerical Columns)
#To understand relationships among all numeric variables at once.
plt.figure(figsize=(8,6))
sns.heatmap(
    df[['Tenure_Months','Quantity','Avg_Price','Delivery_Charges']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()
```

Insight:
Identifies strong or weak relationships between numerical features such as price, quantity, tenure, and delivery charges.

FINAL CONCULSION
Bivariate and multivariate analysis helps uncover relationships between customer demographics, purchasing behavior, pricing, tenure, and coupon usage. These analyses provide deeper insights into customer spending patterns, loyalty impact, and promotional effectiveness, supporting better business and marketing decisions.
